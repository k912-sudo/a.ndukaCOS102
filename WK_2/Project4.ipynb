```python
def calculate_atr():
    print("--- Izifin Technology ATR Calculator ---")
    
    try:
        experience = float(input("Enter years of experience: "))
        age = int(input("Enter age of staff: "))

        if experience > 25 and age >= 55:
            atr = 5600000
        elif experience > 20 and age >= 45:
            atr = 4480000
        elif experience > 10 and age >= 35:
            atr = 1500000
        elif experience < 10 and age < 35:
            atr = 550000
        else:
            atr = 0  # Default if no criteria are met
            print("Staff does not fall into a specific tax bracket.")

        if atr > 0:
            print(f"The Annual Tax Revenue (ATR) is: N{atr:,}")
            
    except ValueError:
        print("Invalid input. Please enter numerical values.")

# Run the calculator
calculate_atr()
```
