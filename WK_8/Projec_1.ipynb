```python
import tkinter as tk
from tkinter import messagebox

# --- OOP Structure ---

class Division:
    """Base Class (Inheritance)"""
    def __init__(self, name):
        self.name = name

    def get_services(self):
        """Placeholder for Polymorphism"""
        return []

class RetailBanking(Division):
    def get_services(self):
        return [
            "Lines of credit", "Investment management and accounts",
            "Insurance", "Retirement and education accounts",
            "Loans and mortgages", "Checking and saving"
        ]

class GlobalBanking(Division):
    def get_services(self):
        return [
            "Multi-currency management services and products",
            "Foreign currency accounts", "Foreign currency credit cards",
            "Transborder advisory services", "Liquidity management"
        ]

class CommercialBanking(Division):
    def get_services(self):
        return [
            "Lines of credit", "Investment management and accounts",
            "Insurance", "Advisory services"
        ]

# --- GUI Logic ---

class BankingApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Banking Division Service Portal")
        self.root.geometry("400x350")

        # UI Elements
        tk.Label(root, text="Employee Name:").pack(pady=5)
        self.name_entry = tk.Entry(root)
        self.name_entry.pack(pady=5)

        tk.Label(root, text="Select Division:").pack(pady=5)
        self.division_var = tk.StringVar(value="Retail Banking")
        divisions = ["Retail Banking", "Global Banking", "Commercial Banking"]
        self.dropdown = tk.OptionMenu(root, self.division_var, *divisions)
        self.dropdown.pack(pady=5)

        tk.Button(root, text="Display Services", command=self.show_info).pack(pady=20)

    def show_info(self):
        emp_name = self.name_entry.get()
        div_choice = self.division_var.get()

        if not emp_name:
            messagebox.showwarning("Input Error", "Please enter an employee name.")
            return

        # Polymorphism in action: 
        # Selecting the correct object based on input
        if div_choice == "Retail Banking":
            div_obj = RetailBanking("Retail")
        elif div_choice == "Global Banking":
            div_obj = GlobalBanking("Global")
        else:
            div_obj = CommercialBanking("Commercial")

        # Fetching services via polymorphic method call
        services = "\n".join([f"- {s}" for s in div_obj.get_services()])
        result_text = f"Employee: {emp_name}\nDivision: {div_choice}\n\nServices rendered:\n{services}"
        
        messagebox.showinfo("Division Services", result_text)

if __name__ == "__main__":
    root = tk.Tk()
    app = BankingApp(root)
    root.mainloop()
```
