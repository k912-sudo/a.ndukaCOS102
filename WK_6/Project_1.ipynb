```python
import tkinter as tk
from tkinter import messagebox
import random

class Employee:
    def __init__(self, name):
        self.name = name
        self.employees = [
            "Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", 
            "Andrew Ugwu", "Stella Mankinde", "Jane Akibo", "Ago James", 
            "Michell Taiwo", "Abraham Jones", "Nicole Anide", "Kosi Korso", 
            "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"
        ]
        self.tasks = [
            "Loading", "Transporting", "Reveiwing Orders", 
            "Customer Service", "Delivering Items"
        ]

    def check_employee(self):
        """Verifies if the name exists in the employee list."""
        if self.name in self.employees:
            self.take_attendance()
            self.assign_task()
        else:
            self.refuse_access()

    def take_attendance(self):
        """Logic for recording attendance."""
        print(f"Attendance recorded for: {self.name}")

    def assign_task(self):
        """Randomly assigns a task and displays a success message."""
        task_index = random.randint(0, len(self.tasks) - 1)
        selected_task = self.tasks[task_index]
        messagebox.showinfo("Access Granted", 
                            f"Welcome, {self.name}!\n"
                            f"Attendance taken successfully.\n"
                            f"Your task today: {selected_task}")

    def refuse_access(self):
        """Displays an error message for unauthorized users."""
        messagebox.showerror("Access Denied", "Identity not found. Access refused.")

# --- GUI Setup ---

def submit():
    user_name = entry.get().strip()
    if not user_name:
        messagebox.showwarning("Input Error", "Please enter a name.")
        return
    
    # Create an instance of the Employee class
    session = Employee(user_name)
    session.check_employee()

# Initialize the main window
root = tk.Tk()
root.title("Logistics Employee Verification")
root.geometry("400x200")

# UI Elements
label = tk.Label(root, text="Enter Employee Name:", font=("Arial", 12))
label.pack(pady=10)

entry = tk.Entry(root, width=30)
entry.pack(pady=5)

submit_btn = tk.Button(root, text="Verify Identity", command=submit, bg="blue", fg="white")
submit_btn.pack(pady=20)

root.mainloop()
```
