```python
import tkinter as tk
from tkinter import messagebox

class DeliveryService:
    def __init__(self, master):
        self.master = master
        master.title("Delivery Charge Calculator")
        master.geometry("400x300")

        # UI Elements
        self.label_location = tk.Label(master, text="Select Location:")
        self.label_location.pack(pady=5)

        self.location_var = tk.StringVar(value="PAU")
        self.radio_pau = tk.Radiobutton(master, text="PAU", variable=self.location_var, value="PAU")
        self.radio_epe = tk.Radiobutton(master, text="Epe", variable=self.location_var, value="Epe")
        self.radio_pau.pack()
        self.radio_epe.pack()

        self.label_weight = tk.Label(master, text="Enter Package Weight (kg):")
        self.label_weight.pack(pady=5)

        self.entry_weight = tk.Entry(master)
        self.entry_weight.pack(pady=5)

        self.calculate_btn = tk.Button(master, text="Calculate Fee", command=self.calculate_fee)
        self.calculate_btn.pack(pady=20)

    def calculate_fee(self):
        try:
            weight = float(self.entry_weight.get())
            location = self.location_var.get()
            fee = 0

            if location == "PAU":
                if weight >= 10:
                    fee = 2000
                else:
                    fee = 1500
            elif location == "Epe":
                if weight >= 10:
                    fee = 5000
                else:
                    fee = 4000

            messagebox.showinfo("Delivery Quote", f"The delivery charge to {location} is: N{fee}")
            
        except ValueError:
            messagebox.showerror("Input Error", "Please enter a valid numerical weight.")

# Main execution
if __name__ == "__main__":
    root = tk.Tk()
    app = DeliveryService(root)
    root.mainloop()
```
